```python
import torch
import numpy as np
```

```python
data = [[2,3],[4,5]]
tensor = torch.tensor(data,dtype=torch.float32)
print(tensor)
```

```python
arr = np.arange(1,10,2)
print(type(arr))
print("simple Numpy array : ",arr)
t1 = torch.from_numpy(arr)
print(type(t1))
print("simple tensor  : ",t1)
```

```python
print(f"ones tensor :\n {torch.ones((2,2))}")
print(f"ones like : \n {torch.ones_like(t1)}")
print(f"random tensor : \n {torch.rand((2,3),dtype=torch.float32)}")
print(f"")
```

Attributes of a Tensor

```python
tensor = torch.rand((2,3))
print("Datatype of Tensor : ",tensor.dtype)
print("Shape of Tensor : ",tensor.shape)
print("Device of Tensor : ",tensor.device)
```

Operation on tensor

```python
tensor = torch.randint(low=0,high=10,size=(4,4),dtype=torch.float32)
print("Original Tensor : \n",tensor)

print("Row-1 : ",tensor[0])
print("column-1 : ",tensor[:,0] )
print("last column : ",tensor[...,-1])
print("Last row : ",tensor[-1])


```

```python
t1 = torch.randint(1,5,(2,2))
t2 = torch.randint(6,9,(2,2))
print(f"Tensor-1 : \n {t1} \nTensor-2 : \n{t2}")
t3 = torch.cat([t1,t2],dim=1) # dim = 1 is for row and 0 is for column
print("tensor-3")
print(t3)
```

```python
t1 = torch.randint(1,5,(2,2),dtype=torch.float32)
t2 = torch.randint(1,5,(2,2),dtype=torch.float32)
print(f"Tensor-1 : \n{t1}")
print(f"Tensor-2 : \n{t2}")
y1 = t2 @ t1
y1 = t1.matmul(t2)
print("Matrix multiplication : \n",y1)
y2 = t1 * t2
print("Elementwise multiplication : \n",y2)
```
