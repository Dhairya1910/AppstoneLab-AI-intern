```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN,Embedding
from tensorflow.keras.datasets import imdb
from tensorflow.keras.utils import pad_sequences
```

```python
(X_train,Y_train),(X_test,Y_test) = imdb.load_data()
```

```python
# to have same shapes according to max len sentence
X_train = pad_sequences(X_train,padding='post',maxlen=1000)
X_test = pad_sequences(X_test,padding='post',maxlen=1000)
```

# Using Simple Interger Encoding method

```python
model = Sequential()
model.add(SimpleRNN(32,input_shape=(X_train[0].shape[0],1),return_sequences=False))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
hist = model.fit(X_train,Y_train,epochs=5,validation_data=(X_test,Y_test))
```

# Using Embbeding Method

```python
model2 = Sequential()
model2.add(Embedding(input_dim=10000, output_dim=32, input_length=1000))
model2.add(SimpleRNN(32,return_sequences=False))
model2.add(Dense(1,activation="sigmoid"))

model2.compile(
    optimizer='adam',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

```python
hist = model2.fit(X_train,Y_train,epochs=5,validation_data = (X_test,Y_test))
```

```python
import tensorflow
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,SimpleRNN,Input,Embedding
from tensorflow.keras.datasets import imdb
from tensorflow.keras.utils import pad_sequences
import numpy as np
```

```python
(train_X,train_Y),(test_X,test_Y) = imdb.load_data()
```

```python
unique_tokens = lambda reviews: set(token for review in reviews for token in review)
all_unique_tokens = unique_tokens(train_X)
print(len(all_unique_tokens))
```

```python
print(f"Shape of Training Data")
print(train_X.shape)
print(train_Y.shape)
print(f"Shape of Testing Data")
print(test_X.shape)
print(test_Y.shape)
```

```python
model = Sequential()
model.add(Embedding(input_dim=20000,output_dim=32,input_length=10000))
model.add(SimpleRNN(32,return_sequences=False))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(
    optimizer = "adam",
    loss = "binary_crossentropy",
    metrics = ['accuracy']
)
```

```python
print(type(train_X), train_X.dtype, train_X.shape)
print(type(train_Y), train_Y.dtype, train_Y.shape)
```

```python
train_X_pradded = pad_sequences(train_X,maxlen=1000,padding='post')
test_X_pradded = pad_sequences(test_X,maxlen=1000,padding='post')
```

```python
train_X = np.array(train_X_pradded,dtype=np.int32)
test_X = np.array(test_X_pradded,dtype=np.int32)
```

```python
hist = model.fit(
    train_X,
    train_Y,
    batch_size=64,
    epochs=5,
    validation_data = (test_X,test_Y)
)
```
