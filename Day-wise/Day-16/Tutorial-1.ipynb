# Exploring NLTK

```python
import nltk 
from nltk import pos_tag
from nltk.tokenize import word_tokenize,TweetTokenizer,WordPunctTokenizer,TreebankWordTokenizer,PunktSentenceTokenizer
from nltk.tokenize.punkt import PunktParameters
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.corpus import stopwords
import inflect
import re
import string
```

**1.TweetTokenizer :** 

**preserve_case (bool)** – Flag indicating whether to preserve the casing (capitalisation) of text used in the tokenize method. Defaults to True.

**reduce_len (bool)** – Flag indicating whether to replace repeated character sequences of length 3 or greater with sequences of length 3. Defaults to False.

**strip_handles (bool)** – Flag indicating whether to remove Twitter handles of text used in the tokenize method. Defaults to **False.

**match_phone_numbers (bool)** – Flag indicating whether the tokenize method should look for phone numbers. Defaults to True.

```python
text = "Hello my name is @Dhairya , I am an AI Engineer my enrollment is 206370307021"
tokenizer = TweetTokenizer(strip_handles=True,reduce_len=True,match_phone_numbers=True)
tokens = tokenizer.tokenize(text)
print(tokens)
```

Custom Sentence Tokeninzer using PunktSentenceTokenizer

```python
text = """Peter parker modified decay rate algo. and made a mistake in the formula, resulting in Dr. Lizard. 
But it was due to greedy indent of Dr. Connors."""
puntk_params = PunktParameters()
puntk_params.abbrev_types = {"algo"}

custom_tokenizer = PunktSentenceTokenizer(puntk_params)

sentence = custom_tokenizer.tokenize(text)
print(sentence)
```

Text Preprocessing using NLTK

```python
# 1. Converting into lower
text = "Text processing is a key component of Natural Language Processing (NLP). It helps us clean and convert raw text data into a format suitable for analysis and machine learning. Dr. Rohit Sharma has scored total of 50 Centuries in International Cricket and has most number of Sixes ie 386 in History of Cricket"
p = inflect.engine()
text = text.lower()

#2. remove number from the text and replace with letters 
temp = text.split()
s2 = []

for word in temp:
    if word.isdigit():
        s2.append(p.number_to_words(word))
    else:
        s2.append(word)
res = " ".join(s2)
# print(res)
text = res

# 3. remove punctuation
trans = str.maketrans("","",string.punctuation)
text = text.translate(trans)
# print(text)

#4. Remove stopwords 
tokens = word_tokenize(text)
tokenized_text = [word for word in tokens if word not in stopwords.words('english')]
# print(tokenized_text)

# 5. Stemming 
stemmer = PorterStemmer()
stemmed_text = [stemmer.stem(word) for word in tokenized_text]

# print(stemmed_text)

# 6. lemmatizer 
lemmatizer = WordNetLemmatizer()
lemmatized_text = [lemmatizer.lemmatize(word) for word in tokenized_text]
# print(lemmatized_text)

# 7. POS tagging 
tags = pos_tag(lemmatized_text)
print(tags)
```

N-Gram modeling

```python
from nltk import trigrams 
from nltk.corpus import reuters 
from collections import defaultdict 

words = word_tokenize(" ".join(reuters.words()))
tri_gram = list(trigrams(words))
model = defaultdict(lambda:defaultdict(lambda:0))

for word1,word2,word3 in tri_gram:
    model[(word1,word2)][word3] = model[(word1,word2)][word3] + 1 

for word1_word2 in model:
    total_count = float(sum(model[word1_word2].values()))
    for word3 in model[word1_word2]:
        model[word1_word2][word3] /= total_count 

def pred_next_word(w1,w2):
    next_word_probs = model[w1, w2]
    if next_word_probs:
        return max(next_word_probs, key=next_word_probs.get)
    else:
        return "No prediction available"

print("Next Word : ",pred_next_word("World","economy"))
```
