# Text Splitter in Langchain 

1. Length based splitter

```python
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.document_loaders import DirectoryLoader,PyPDFLoader
```

```python
loader = DirectoryLoader(
    path = r"d:\AppstoneLab-AI-intern\Concept-Wise\Gen-AI\DocumentLoaders\Docs",
    glob= "*.pdf",
    loader_cls= PyPDFLoader
)

docs = loader.load()

docs

splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=0,
    separator=" "
)

results = splitter.split_documents(docs)
print(results[0].page_content)
```

2. Text-Structure based text splitter

in this type of splitter main focus on the structure of the given text.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language
```

```python
loader = DirectoryLoader(
    path = r"d:\AppstoneLab-AI-intern\Concept-Wise\Gen-AI\DocumentLoaders\Docs",
    glob= "*.pdf",
    loader_cls= PyPDFLoader
)

docs = loader.load()

splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=0
)

results = splitter.split_documents(docs)
print(results[0].page_content)
```

3. Document based text splitter

```python
code = """
class Helloworld:
    def __init__(self,x):
        self.x = x 
    def hellu(self):
        return self.x
"""


splitter = RecursiveCharacterTextSplitter.from_language(
    chunk_size=50,
    language = Language.PYTHON,
    chunk_overlap=0
)

results = splitter.split_text(code)
print(len(results))
print(results[0])
print(results[1])
print(results[2])
```
