# Document Loaders in Langchain

```python
from langchain_community.document_loaders import TextLoader 
from langchain_core.output_parsers import StrOutputParser 
from langchain_core.prompts import PromptTemplate 
from langchain_mistralai import ChatMistralAI 
from dotenv import load_dotenv 
```

```python
if load_dotenv():
    print("api keys verified") 

model = ChatMistralAI(
    model = 'mistral-medium-2508',
    temperature = 1
)
print("model successfully loaded.")
```

```python
content = TextLoader(
    file_path= "Rohit.txt"
)
docs = content.load()
docs[0]
```

```python
parser = StrOutputParser()

prompt = PromptTemplate(
    template = "sumarize the following {content} in 100 words",
    input_variables = ['content']
)
print(prompt)
```

```python
chain = prompt | model | parser 
res = chain.invoke(
    {
        "content" : docs[0].page_content
    }
)
res
```

# Web-Based Loader

```python
from langchain_community.document_loaders import WebBaseLoader 
```

```python
loader = WebBaseLoader(
    "https://www.amazon.in/Sony-PS5-Ghost-of-Yotei/dp/B0F6VH981P/ref=sr_1_3?sr=8-3"
)

docs = loader.load()
```

```python
prompt = PromptTemplate(
    template = "what is the price of product we are viewing refer : {text}",
    input_variables = ['text']
)
```

```python
chain = prompt | model | parser 
res = chain.invoke(
    {
        "text" : docs[0].page_content
    }
)
```

```python
print(res)
```
