# Custom tools

```python
from langchain.tools import tool 
```

```python
@tool 
def multiply(a:int,b:int) -> int:
    """
    function performs multiplication of two number 

    **Arguments**
    a = int 
    b = int
    """

    return a * b 
```

```python
out = multiply.invoke({
    "a" : 10,
    "b" : 30
})
print(out)
```

# Creating Structured tools 

```python
from pydantic import BaseModel,Field 
from typing import Literal
```

```python
class WeatherInput(BaseModel):
    location : str = Field(description="nome of the location")
    units : Literal["celsius","fahrenheit"] = Field(description="Temperature unit")
    forcast : str = Field(description="5-days forcast")
```

```python
@tool(args_schema=WeatherInput)
def get_weather(location:str,units:str,forcast:str) -> str:
    """Get current weather and optional forecast."""
    temp = 22 if units == "celsius" else 72
    result = f"Current weather in {location} is {temp}{units[0]}"
    return result
```

```python
get_weather.invoke({
    "location": "surat",
    "units": "celsius",
    "forcast": "sunny"
})
```

# Using BaseTools 

```python
from typing import Type
from langchain.tools import BaseTool 
```

```python
class MultiplyInput(BaseModel):
    a : int = Field(description = "The first number to multiply")
    b : int = Field(description = "The second number to multiply")
```

```python
class MultiplyTool(BaseTool):
    name: str = "Multiply"
    description : str = "Multiply two numbers"
    arg_schema : Type[BaseModel] = MultiplyInput 

    def _run(self,a: int,b: int) -> int:
        return a * b
```

```python
multiply_tool = MultiplyTool()
```

```python
multiply_tool.invoke(
    {
        "a" : 10,
        "b" : 10
    }
)
```

# Creating a simple toolkit

```python
from langchain.tools import tool
```

```python
@tool 
def multiply(num1:int,num2:int) -> int:
    """
    This function can be used to multiply two number.
    **Arguments**
    num1 & num2     
    """
    return num1 * num2

@tool 
def Addition(num1:int,num2:int) -> int:
    """
    This Function can be used to add two number.
    **Arguments**
    num1 & num2
    """

    return num1 + num2
```

```python
from langchain_core.tools import BaseToolkit
class MathsToolKit(BaseToolkit):
    def get_tools(self) -> list:
        return [Addition, multiply]
```

# Tool binding 

```python
from langchain_mistralai import ChatMistralAI 
from dotenv import load_dotenv
from langchain.messages import HumanMessage 
load_dotenv()
```

```python
llm = ChatMistralAI(
    model = "mistral-medium-2508"
)
```

```python
@tool 
def multiply(num1:int,num2:int) -> int:
    """
    This function can be used to multiply two number.
    **Arguments**
    num1 & num2     
    """
    return num1 * num2
```

```python
llm_with_tools = llm.bind_tools([multiply])
```

```python
llm_with_tools
```

```python
message = [HumanMessage(content="can you multiply 2 with 60")]
message
```

```python
llm_with_tools
```

```python
result = llm_with_tools.invoke(message)
message.append(result)
```

```python
res = multiply.invoke(result.tool_calls[0])
message.append(res)
```

```python
llm_with_tools.invoke(message)
```

# Currency conversation using tools 

```python
import requests 
from langchain.tools import tool 
```

```python
@tool 
def get_conversion_factor(base_code:float,target_code:float)->float:
    url = f"https://v6.exchangerate-api.com/v6/e3e8acaae41517abebc43ae8/pair/{base_code}/{target_code}"
    response = requests.get(url)
    return response.json()['cov']
```
