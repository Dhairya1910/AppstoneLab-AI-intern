```python
from langchain_mistralai import MistralAIEmbeddings 
from langchain_chroma import Chroma 
from langchain_core.documents import Document
from dotenv import load_dotenv
load_dotenv()
```

```python
embedding_model = MistralAIEmbeddings(
    model = "mistral-embed"
)
```

```python
doc1 = Document(
    page_content="Virat kholi is one of the most successful and consistent batsman in IPL history.",
    metadata = {"team":"Royal Challangers Bangalore"}
)

doc2 = Document(
    page_content= "Rohi Sharma is one of Greatest and most succesful white ball batsman, his Aggressive batting and sharp captain mindset.",
    metadata = {"Team":"Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni, famously known as Captain cool, has led Chennai Super kings to win multiple titles.",
    metadata = {"Team":"Chennai Super Kings"}
)
```

```python
vector_store = Chroma(
    collection_name="Cricket_data",
    persist_directory= r"d:\AppstoneLab-AI-intern\ChromaDB",
    embedding_function= embedding_model
)
```

```python
Complete_docs = [doc1,doc2,doc3] 
vector_store.add_documents(Complete_docs)
```

```python
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
# searching a vector store 
vector_store.similarity_search(
    query = "who is the greatest white ball cricketer",
    k=1
)
```

```python
# Searching a vector db with respect to score 
vector_store.similarity_search_with_score(
    query= "who is the greatest white ball cricketer",
)
```

```python
# updating the data based on doc-id's 

updated_doc = Document(
    page_content= "Rohit Sharma is one the greatest white ball cricketer as well as most successfull captain in ODI cricket.",
    metadata = {"team":"Mumbai Indians"}
)

vector_store.update_document(document_id="59ffe58b-2498-4775-9ddc-0586ab6cacb3",document=updated_doc)
```

```python
vector_store.get(
    include=['embeddings','documents','metadatas']
)
```

```python
vector_store.similarity_search_with_score(
    query="who is the best captain for Odi cricket",
    k =2 
)
```

```python
# Deleting Documents 
vector_store.delete(
    ids = [
        'cfbec8ff-587d-4963-bd20-0a1fad09426c',
        '979e2cca-3406-42ac-bcd5-ebe8e20f267c',
        '42e33c3c-5b3a-44af-856f-f95dcdeda45e'
        ]
)

vector_store.get(
    include=['embeddings','documents','metadatas']
)
```

```python
# search using embedding vector
vector_store.similarity_search_by_vector(
    embedding= embedding_model.embed_query("who is the best captain for Odi cricket"),
    k = 1
)
```

3. Multi-Query retriever

```python
from langchain_classic.retrievers.multi_query import MultiQueryRetriever 
from langchain_mistralai import ChatMistralAI
```

```python
docs = [
    # Contradictory facts
    Document(page_content="The Olympics are held every 4 years.", metadata={"source": "wiki", "year": 2024}),
    Document(page_content="The Olympics are held every 2 years.", metadata={"source": "blog", "year": 2023}),

    # Ambiguous entities
    Document(page_content="Apple was founded in 1976 and is known for its fruit processing plants.", metadata={"source": "article", "year": 2021}),
    Document(page_content="Apple is the most valuable tech company in the world.", metadata={"source": "news", "year": 2023}),

    # Overlapping but conflicting dates
    Document(page_content="World War II ended in 1945 after the bombing of Hiroshima.", metadata={"source": "history", "year": 2020}),
    Document(page_content="World War II ended in 1944 with the D-Day invasion.", metadata={"source": "textbook", "year": 2019}),

    # Same topic, totally different context
    Document(page_content="Python is a snake found in tropical regions of Africa and Asia.", metadata={"source": "nature", "year": 2022}),
    Document(page_content="Python is the most popular programming language for AI and data science.", metadata={"source": "tech", "year": 2023}),
    Document(page_content="Python was created by Guido van Rossum in 1989 and is named after a TV show.", metadata={"source": "wiki", "year": 2021}),

    # Partial truths
    Document(page_content="Einstein never passed his mathematics exams in school.", metadata={"source": "myth_site", "year": 2020}),
    Document(page_content="Einstein excelled in mathematics and physics from a very young age.", metadata={"source": "biography", "year": 2018}),

    # Misleading statistics
    Document(page_content="90% of car accidents happen within 10 miles of home, making highways safer.", metadata={"source": "blog", "year": 2022}),
    Document(page_content="Highways are the most dangerous roads due to high speeds and traffic volume.", metadata={"source": "safety_report", "year": 2023}),

    # Vague and unhelpful
    Document(page_content="Some scientists believe climate change may or may not affect weather patterns in certain regions.", metadata={"source": "vague_news", "year": 2021}),
    Document(page_content="Climate change is definitively causing more extreme weather events globally.", metadata={"source": "nasa", "year": 2023}),

    # Temporal confusion
    Document(page_content="The iPhone was first released in 2007 and revolutionized the smartphone industry.", metadata={"source": "tech_history", "year": 2020}),
    Document(page_content="Smartphones became popular around 2003 when Apple launched the first iPhone.", metadata={"source": "incorrect_blog", "year": 2019}),

    # Completely off-topic noise
    Document(page_content="The best way to cook pasta is to boil it for exactly 8 minutes with salt.", metadata={"source": "recipe", "year": 2022}),
    Document(page_content="Pasta should never be cooked with salt as it damages the pot.", metadata={"source": "myth_cooking", "year": 2021}),

    # Mixed language confusion
    Document(page_content="The Eiffel Tower is located in Berlin, Germany and was built in 1889.", metadata={"source": "wrong_travel", "year": 2020}),
    Document(page_content="The Eiffel Tower is located in Paris, France and stands 330 meters tall.", metadata={"source": "travel_guide", "year": 2022}),
]

# add these to your existing docs list

docs += [
    # Self-contradicting single documents
    Document(page_content="Water boils at 100°C, which is why it remains liquid at 100°C under normal conditions.", metadata={"source": "science_blog", "year": 2022}),
    Document(page_content="The sun rises in the west and sets in the east, giving us beautiful morning sunsets.", metadata={"source": "nature_blog", "year": 2021}),

    # Numbers that sound plausible but conflict
    Document(page_content="The Great Wall of China is approximately 13,000 miles long.", metadata={"source": "travel_wiki", "year": 2022}),
    Document(page_content="The Great Wall of China stretches for about 1,500 miles across northern China.", metadata={"source": "encyclopedia", "year": 2020}),
    Document(page_content="The Great Wall of China is 21,196 kilometers long according to recent surveys.", metadata={"source": "china_gov", "year": 2023}),

    # Same person, conflicting biographical facts
    Document(page_content="Nikola Tesla was born in 1856 in Serbia and died penniless in New York in 1943.", metadata={"source": "biography", "year": 2021}),
    Document(page_content="Nikola Tesla was born in 1843 in Croatia and lived a wealthy life until his death in 1950.", metadata={"source": "alt_history", "year": 2020}),
    Document(page_content="Tesla never held any patents and gave all his inventions freely to the public.", metadata={"source": "myth_site", "year": 2022}),

    # Cause and effect reversed
    Document(page_content="Exercise causes weight loss because people who lose weight tend to exercise more.", metadata={"source": "health_blog", "year": 2022}),
    Document(page_content="Sleeping more causes tiredness, as studies show tired people sleep longer hours.", metadata={"source": "wellness_site", "year": 2021}),

    # Subtle unit confusion
    Document(page_content="The moon is approximately 384,400 kilometers away from Earth.", metadata={"source": "nasa", "year": 2023}),
    Document(page_content="The moon is approximately 384,400 miles away from Earth, making it very close.", metadata={"source": "space_blog", "year": 2022}),

    # Famous quotes misattributed
    Document(page_content="Albert Einstein once said, 'The definition of insanity is doing the same thing and expecting different results.'", metadata={"source": "quote_site", "year": 2021}),
    Document(page_content="The quote about insanity and repetition has no verified attribution to Einstein and likely originated elsewhere.", metadata={"source": "fact_check", "year": 2022}),
    Document(page_content="Benjamin Franklin coined the phrase about insanity in his 1752 letters to the Continental Congress.", metadata={"source": "history_blog", "year": 2020}),

    # Plausible but completely fabricated studies
    Document(page_content="A Harvard study in 2018 found that people who drink 5 cups of coffee daily live 10 years longer.", metadata={"source": "health_news", "year": 2021}),
    Document(page_content="A Stanford study in 2019 found that any coffee consumption above 2 cups significantly reduces lifespan.", metadata={"source": "medical_blog", "year": 2022}),

    # Geographic confusion
    Document(page_content="Australia is the largest continent in the world by land area.", metadata={"source": "geography_blog", "year": 2021}),
    Document(page_content="Antarctica is considered both a continent and an ocean by most geographers.", metadata={"source": "edu_site", "year": 2020}),
    Document(page_content="Asia is the largest continent, covering about 44.5 million square kilometers.", metadata={"source": "encyclopedia", "year": 2023}),

    # Technology facts mixed up
    Document(page_content="The first computer was built by Bill Gates in 1975 in his Harvard dorm room.", metadata={"source": "tech_myth", "year": 2020}),
    Document(page_content="ENIAC, built in 1945, is widely considered the first general-purpose electronic computer.", metadata={"source": "tech_history", "year": 2022}),
    Document(page_content="Charles Babbage built the first fully functional computer in 1837 called the Difference Engine.", metadata={"source": "history_site", "year": 2021}),

    # Medical misinformation mixed with facts
    Document(page_content="Humans only use 10% of their brain, leaving 90% untapped potential.", metadata={"source": "self_help", "year": 2021}),
    Document(page_content="Brain scans show that virtually all parts of the brain have some function and are active at various times.", metadata={"source": "neuroscience_journal", "year": 2023}),
    Document(page_content="The 10% brain myth was first proposed by Einstein himself in a 1929 interview.", metadata={"source": "myth_blog", "year": 2020}),

    # Recycled context with swapped subjects
    Document(page_content="Leonardo da Vinci painted the Sistine Chapel ceiling between 1508 and 1512.", metadata={"source": "art_blog", "year": 2021}),
    Document(page_content="Michelangelo painted the Sistine Chapel ceiling, a work often mistakenly credited to da Vinci.", metadata={"source": "art_history", "year": 2022}),
    Document(page_content="The Sistine Chapel ceiling was painted by Raphael under commission from Pope Julius II.", metadata={"source": "wrong_encyclopedia", "year": 2020}),

    # Circular reasoning documents
    Document(page_content="Gold is valuable because people are willing to pay a lot for it.", metadata={"source": "economics_blog", "year": 2022}),
    Document(page_content="People pay a lot for gold because it is inherently valuable as a precious metal.", metadata={"source": "finance_site", "year": 2021}),

    # Completely irrelevant to anything else in the corpus
    Document(page_content="The ideal temperature for storing red wine is between 12 and 18 degrees Celsius.", metadata={"source": "wine_guide", "year": 2022}),
    Document(page_content="Red wine should always be stored at room temperature, which is around 22 degrees Celsius.", metadata={"source": "wine_myth", "year": 2021}),
    Document(page_content="All wine regardless of type should be stored in the freezer to preserve freshness.", metadata={"source": "bad_advice", "year": 2020}),
]
```

```python
Vector_store = Chroma.from_documents(
    embedding = embedding_model,
    documents = docs,
)
```

```python
similarity_retriever = Vector_store.as_retriever(
    search_type = 'similarity',
    search_kwargs = {
        "k" : 2
    }
)
```

```python
similarity_retriever.invoke(
    "creator of  Sistine Chapel",
)
```

```python
mutli_query_retriever = MultiQueryRetriever.from_llm(
    retriever = Vector_store.as_retriever(search_kwargs={"k":3}),
    llm = ChatMistralAI(model='mistral-medium-2508')
)

res = mutli_query_retriever.invoke(
    "creator of  Sistine Chapel"
)
```

4. Contextual Compression retriever

```python
from langchain_classic.retrievers.contextual_compression import ContextualCompressionRetriever 
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
```

```python
model = ChatMistralAI(
    model = "mistral-medium-2508"
)

compressor = LLMChainExtractor.from_llm(model)
```

```python
compression_retriever = ContextualCompressionRetriever(
    base_retriever = similarity_retriever,
    base_compressor = compressor
)
```

```python
res = compression_retriever.invoke(
    "creator of  Sistine Chapel"
)
```
