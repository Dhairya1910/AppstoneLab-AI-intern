# Runnable's in Langchain

```python
from langchain_mistralai import ChatMistralAI 
from langchain_core.output_parsers import StrOutputParser 
from dotenv import load_dotenv 
from langchain_core.runnables import RunnableSequence 
from langchain_core.prompts import PromptTemplate 
```

```python
if load_dotenv():
    print("Api keys fetch successful")

model = ChatMistralAI(
    model = "mistral-medium-2508",
    temperature = 1,
)
print("Model created successfully")
```

```python
template_1 = PromptTemplate(
    template = "tell me a fact on {topic}",
    input_variable = ['topic']
)

template_2 = PromptTemplate(
    template = "explain the {fact}.",
    input_variable = ['fact']
)
```

```python
parser = StrOutputParser()
```

```python
chain = RunnableSequence(template_1,model,parser,template_2,model,parser)
res = chain.invoke({
    "topic" : "cricket"
})
res
```

```python
print(res)
```

# Parallel Runnable 

```python
from langchain_core.runnables import RunnableParallel
```

```python
prompt_1 = PromptTemplate(
    template = "based on {game_name} generate a small summary",
    input_variable = ['game_name']
)

prompt_2 = PromptTemplate(
    template = "based on {game_name} show its system requirements",
    input_variable = ['game_name']
)
```

```python
chain = RunnableParallel(
    {
        "game_title" : RunnableSequence(prompt_1,model,parser),
        "system_requirements" : RunnableSequence(prompt_2,model,parser)
    }
)
```

```python
res = chain.invoke(
    {
        "game_name" : "Total Overdose"
    }
)
```

```python
print(res['game_title'])
print("-"*100)
print(res['system_requirements'])
```

# Runnable Passthrough

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
prompt_1 = PromptTemplate.from_template(
    template = "tell me a random fact about : {person}",
    input_variable = ['person']
)
```

```python
parallel_chain = RunnableParallel(
    {
        "original_prompt" : RunnablePassthrough(),
        "modifed_prompt" : RunnableSequence(prompt_1,model,parser)
    }
)

res = parallel_chain.invoke(
    {
        "person" : "Spiderman"
    }
)
res
```

```python
print(f'Output of Passthrough : {res['original_prompt']}')
print(f'{res['modifed_prompt']}')
```

# Runnable Lambda

```python
from langchain_core.runnables import RunnableLambda 
```

```python
prompt = PromptTemplate(
    template = "tell me a fact about {person}",
    input_variable = ['person']
)
```

```python
def word_count(x):
    return len(x.split())
```

```python
output_chain = RunnableSequence(prompt,model,parser)

parallel_chain = RunnableParallel(
    {
        "fact" : RunnablePassthrough(),
        "count" : RunnableLambda(word_count)
    }
)

final_chain = RunnableSequence(output_chain,parallel_chain)
res = final_chain.invoke(
    {
        "person" : "Rohit Sharma"
    }
)
```

```python
print(f"Fact : {res['fact']}")
print("--"*100)
print(f" total Number of words : {res['count']}")
```

# Runnable Branch

```python
from langchain_core.runnables import RunnableBranch 
from langchain_core.output_parsers import StrOutputParser,PydanticOutputParser
from pydantic import BaseModel,Field
```

```python
class fetch_playing_style(BaseModel):
    PlayerName : str = Field(description="Name of the cricketer")
    Age : int = Field(description="age of the cricketer")
    PlayingStyle  : str = Field(description="Playing style of cricketer in one or two words")
```

```python
pydantic_parser = PydanticOutputParser(pydantic_object=fetch_playing_style)
format_instruction = pydantic_parser.get_format_instructions()
```

```python
prompt = PromptTemplate.from_template(
    template = "Suggest me one single cricketer playing for {country} and generate very small summary about the cricketers that includes there Country and playing style,age.",
    input_variable = ['country'] 
)

selection_prompt = PromptTemplate(
    template = "Based on {summary} \n {format_instruction}",
    input_variables = ['summary'], 
    partial_variables = {
        "format_instruction": pydantic_parser.get_format_instructions()
    }
)
Conditional_prompt = PromptTemplate.from_template(
    template = "describe the playstyle of {player} in his early days.",
)
```

```python
init_chain = RunnableSequence(prompt,model,parser)

middle_chain = RunnableParallel(
        {
        "original" : RunnablePassthrough(),
        "Input" : RunnableSequence(selection_prompt,model,pydantic_parser)
        }
)

Conditional_chain = RunnableBranch(
    (
        lambda x: int(x["Input"].Age) >= 35,
        RunnableSequence(
            RunnableLambda(lambda x: {"player": x["Input"].PlayerName}),
            Conditional_prompt,
            model,
            parser,
        ),
    ),
    RunnableLambda(lambda x:  x["Input"].PlayerName +" is below age here is his summary : \n" + x['original']),
)
```

```python
final_chain = RunnableSequence(init_chain,middle_chain,Conditional_chain)
res = final_chain.invoke(
    {
        "country" : "India"
    }
    
)
```

```python
print(res)
```
