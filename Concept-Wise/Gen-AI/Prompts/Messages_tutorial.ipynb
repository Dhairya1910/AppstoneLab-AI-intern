# Working with ChatPromptTemplate and Messages in Langchain

```python
from langchain_mistralai import ChatMistralAI 
from langchain_core.prompts import ChatPromptTemplate 
from langchain_core.messages import HumanMessage 
from dotenv import load_dotenv 
```

```python
if load_dotenv():
    print("Api keys verified")

model = ChatMistralAI(
    model = "mistral-medium-2508",
    temperature = 0.7,
    max_tokens = 500
)

print("model created successfully")
```

```python
# creating a ChatPromptTemplate 

template = ChatPromptTemplate([
    ('system','You are {domain} expert, you have in depth knowledge about the {domain}.'),
    ('user','which cricket ground is most difficult to bat.'),
])

prompt = template.invoke(
    {
        "domain":'cricket',
    }
)
```

```python
prompt
```

```python
response = model.invoke(prompt)
print(response.content)
```

# Using MessageHolder to retrive old chats from a txt file

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder 
```

**MessagePlaceholder helps to insert old chats from files,databases within the ongoing converation between the model and Human**

```python
template = ChatPromptTemplate([
    ('system','you are a super helpful and sarcastic assistant'),
    MessagesPlaceholder(variable_name="chat_history"), # this line of code will reload the conversation from old chats and insert them between an ongoing converation
    ('user','do you remember my fav cricketer')
])

chat_history = []

with open('chat_history.txt') as f:
    chat_history.extend(f.readlines())


prompt = template.invoke(
    {
        "chat_history" : chat_history
    }
)
```

```python
prompt
```

```python
response = model.invoke(prompt)
print(response.content)
```
