# Agent using a System prompt

```python
from langchain_mistralai import ChatMistralAI 
from langchain.agents import create_agent
from dotenv import load_dotenv 
from langchain.messages import HumanMessage
```

```python
system_prompt = """
You are an outstanding competitive programmer.
please keep the below structure.

input : question asked by user.
method : which method you used for solving the problem
output : only code.
time complexity : time complexity of the approch used.
"""
```

```python
load_dotenv()
model = ChatMistralAI(
    model = 'mistral-medium-2508',
)
print("model loaded Successfully.")
```

```python
coding_agent = create_agent(
    model = model,
    system_prompt=system_prompt
)
print("Agent successfully created.")
```

```python
Question = input("enter your coding problem : ")
print("-"*100)
for chunk in model.stream([HumanMessage(content = Question)]):
    if chunk.content:
        print(chunk.content, end="", flush=True)
print("\n")
print("-"*100)
```

# Intergrating Tools into an Agent

**Tools are user-defined function that can be accessed by an Agent.**

```python
from langchain_openrouter import ChatOpenRouter
from langchain.agents import create_agent 
from langchain.tools import tool
from langchain.messages import HumanMessage 
from dotenv import load_dotenv 
from typing import Dict, Any
load_dotenv()
```

```python
model = ChatOpenRouter(
    model = "arcee-ai/trinity-large-preview:free",
)
print("model built successfully.")
```

```python
from tavily import TavilyClient
tavily_client = TavilyClient()
@tool
def Web_search(query : str):
    """search the web for information"""
    return tavily_client.search(query)

Web_search.invoke("who is current chief minister of Gujarat")
```

```python
agent = create_agent(
    model = model,
    tools = [Web_search],
)

Question = "who is current chief minister of Gujarat."

response = agent.invoke(
    {"messages":[Question]}
)
```

```python
response
```

# Agent with Memory

```python
from langgraph.checkpoint.memory import InMemorySaver 

agent = create_agent(
    model = model,
    checkpointer=InMemorySaver()
)
print("model created successfully.")
```

```python
from langchain.messages import HumanMessage 
question = HumanMessage(content="Hello My name is Dhairya and my fav color is Blue.")
config = {"configurable":{"thread_id":1}}

response = agent.invoke(
    {"messages":[question]},
    config
)
```

```python
print(response)
```

```python
question = HumanMessage(content="what is my Name .?")
response = agent.invoke(
    {"messages":[question]},
    config,
)
print(f"User :{question.content}")
print(f"Model response : {response["messages"][-1].content}")
```

# Multi-Modular input into an Agent

```python
from langchain_mistralai import ChatMistralAI
from langchain.agents import create_agent 
from langchain.tools import tool
from langchain.messages import HumanMessage 
from dotenv import load_dotenv 
load_dotenv()
```

```python
from ipywidgets import FileUpload
from IPython.display import display 

uploader = FileUpload(accept = 'jpg',multiple=False)
display(uploader)
```

```python
print(uploader.value)
```

```python
import base64 
uploaded_file = uploader.value[0]
# read uploader content
content_mv = uploaded_file['content']
# convert it into bytes
img_byte = bytes(content_mv)
# encode as per base64
img_base64 = base64.b64encode(img_byte).decode("utf=8")
```

```python
model = ChatMistralAI(
    model = 'mistral-medium-2508',
)

agent = create_agent(
    model = model,
)
```

```python
multiModal_question = HumanMessage(
    content=[
        {"type": "text", "text": "What clothes is this person wearing?"},
        {
            "type": "image_url",
            "image_url": {
                "url": f"data:image/jpeg;base64,{img_base64}"
            }
        }
    ]
)
response = agent.invoke(
    {
        "messages" : [multiModal_question]
    }
)
```
