# Trying Out different tools using LangChain

```python
from langchain_mistralai import ChatMistralAI
from langchain.agents import create_agent
from langchain.tools import tool
from dotenv import load_dotenv
from langchain.messages import HumanMessage  
from tavily import TavilyClient
load_dotenv()
```

```python
@tool 
def Fetch_news(query):
    """ For getting latest news and updates"""
    client = TavilyClient()
    return client.search(query)

# Fetch_news.invoke("who is current odi captain Team India")
```

```python
model = ChatMistralAI(
    model = "mistral-medium-2508",
    temperature = 0.5,
    max_tokens = 500
)

print("model loaded successfully.")

Mistral_agent = create_agent(
    model = model,
    tools = [Fetch_news],
    system_prompt= "You are very Intelligent Assistant,use your tools to assist the user."
)
print("agent loaded successfully.")
```

```python
message = {"messages" : [HumanMessage("who is current team India captain for ODI")]}
for chunk,metadata in Mistral_agent.stream(message,stream_mode="messages"):
    print(chunk.content,end="",flush=True)
```

```python
message = {"messages" : [HumanMessage("who is current captain of the Indian men's ODI team")]}
for chunk,metadata in Mistral_agent.stream(message,stream_mode="messages"):
    if chunk.model_dump()['type'] == 'AIMessageChunk':
        print(chunk.content,end="",flush=True)
```
